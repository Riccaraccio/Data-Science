```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# --- Dataset ---
# Reactor conversion study:
# Two catalysts (A and B) tested across a range of temperatures (300–600 K)
# at two pressures (1 bar and 5 bar).
# Variable measured: conversion (0–1)

np.random.seed(42)
temperatures = np.array([300, 350, 400, 450, 500, 550, 600])

data = {
    'Temperature_K': np.tile(temperatures, 4),
    'Pressure_bar': np.repeat([1, 1, 5, 5], len(temperatures)),
    'Catalyst': np.repeat(['A', 'B', 'A', 'B'], len(temperatures)),
    'Conversion': np.concatenate([
        0.10 + 0.10 * np.arange(len(temperatures)) + np.random.normal(0, 0.02, len(temperatures)),  # A, 1 bar
        0.08 + 0.08 * np.arange(len(temperatures)) + np.random.normal(0, 0.02, len(temperatures)),  # B, 1 bar
        0.20 + 0.11 * np.arange(len(temperatures)) + np.random.normal(0, 0.02, len(temperatures)),  # A, 5 bar
        0.15 + 0.09 * np.arange(len(temperatures)) + np.random.normal(0, 0.02, len(temperatures)),  # B, 5 bar
    ])
}

df = pd.DataFrame(data)
df['Conversion'] = df['Conversion'].clip(0, 1).round(3)

print(df.head(14))
```
