# K-means Clustering Visualisation on Synthetic Data

This notebook demonstrates K-means clustering on synthetic 2D data by:
1. Generating 5 Gaussian clusters using `make_blobs`
2. Applying K-means to identify cluster centres and assignments
3. Visualising the results with colour-coded points and cluster centres

The synthetic dataset consists of 200 points distributed around
5 centres with a standard deviation of 0.6.

## Generate Synthetic Clustered Data

`make_blobs` creates isotropic Gaussian clusters. Here we generate 200 points
around 5 centres with low spread (`cluster_std=0.6`) so the clusters are well-separated.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate synthetic clustered data
X, Y = make_blobs(n_samples=200,    # Total number of points
                 centers=5,         # Number of clusters
                 cluster_std=0.6,   # Cluster spread
                 random_state=0)    # For reproducibility
```

## Perform K-means Clustering

K-means partitions the data into `k` clusters by minimising the within-cluster
sum of squared distances to the cluster centroid.

```python
# Perform K-means clustering
kmeans = KMeans(n_clusters=5)  # Initialize with 5 clusters
kmeans.fit(X)                  # Fit model to data
y_kmeans = kmeans.labels_      # Get cluster assignments
```

## Visualise Results

Each point is coloured by its cluster assignment. The black dots mark the
cluster centres found by K-means.

```python
# Visualize results
plt.scatter(X[:, 0], X[:, 1],     # Plot data points
          c=y_kmeans,             # Color by cluster assignment
          s=50,                   # Point size
          cmap='viridis')         # Color scheme

# Plot cluster centers
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1],  # Plot centers
          c='black',                      # Center color
          s=200,                          # Center point size
          alpha=0.5)                      # Transparency
plt.show()
```
